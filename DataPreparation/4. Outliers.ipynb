### 1. Load Library

```python
import numpy as np
import pandas as pd
```

### 2. Load Data

```python
df = pd.DataFrame({"pendapatan": [3, 4, 5, 4, 6, 5, 4, 50, 4]})
```

### 3.1 Interquartile Range (IQR)

```python
Q1 = df["pendapatan"].quantile(0.25)
Q3 = df["pendapatan"].quantile(0.75)
IQR = Q3 - Q1
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR
```

### 3.1.1 Deteksi

```python
outliers_iqr = df[
    (df["pendapatan"] < batas_bawah) | (df["pendapatan"] > batas_atas)
]
outliers_iqr
```

### 3.1.2. Clipping (memotong ke batas terdekat)

```python
df["pendapatan_clipped"] = df["pendapatan"].clip(
    lower=batas_bawah, upper=batas_atas
)
df
```
