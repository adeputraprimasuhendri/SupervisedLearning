# Simple Random Split vs. Stratified Split

### 1. Load Library

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

### 2. Load Dataset

```python
y = pd.Series([0] * 90 + [1] * 10)
X = pd.DataFrame({"nilai": range(100)})
```

### 3.1 Simple Random Split

```python
X_train_rnd, X_test_rnd, y_train_rnd, y_test_rnd = train_test_split(
    X, y, test_size=0.20, random_state=42, shuffle=True
)
```

```python
print("Distribusi Target - Simple Random Slit Test Set:")
print(y_test_rnd.value_counts(normalize=True))
```

### 3.2 Stratified Split

```python
X_train_str, X_test_str, y_train_str, y_test_str = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
print("Distribusi Target - Stratified Test Set:")
print(y_test_str.value_counts(normalize=True))
```
