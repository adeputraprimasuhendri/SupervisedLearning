### 1. Load Library

```python
import pandas as pd
```

### 2. Load Dataset

```python
data = {
    "kota": [
        " Jakarta ",
        "jkt",
        "JAKARTA",
        "Bandung",
        "Bandung",
        "  surabaya  ",
    ],
    "kategori": [
        "Elektronik",
        "elektronik",
        "ELEKTRONIK",
        "Pakaian",
        "Pakaian",
        "makanan",
    ],
}
df = pd.DataFrame(data)
df
```

### 3. Standardisasi Huruf (Lowercasing) & Pembersihan Spasi

```python
kolom_teks = ["kota", "kategori"]
for col in kolom_teks:
    df[col] = (
        df[col]
        .str.lower() #ubah semua huruf jadi kecil
        .str.strip() #hapus spasi di awal dan akhir string
        .str.replace(r"\s+", " ", regex=True) #ubah spasi ganda di tengah jadi 1 spasi
    )
df
```

### 4. Ketidakkonsistenan Pengetikan / Singkatan

```python
kamus_koreksi = {
    "jkt": "jakarta",
    "bdg": "bandung",
    "sby": "surabaya",
}
df["kota"] = df["kota"].replace(kamus_koreksi)
df
```

### 5. Penghapusan Duplikasi

```python
df_clean = df.drop_duplicates().reset_index(drop=True)
df_clean
```
