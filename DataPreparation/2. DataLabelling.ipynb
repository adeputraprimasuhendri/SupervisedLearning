### 1. Load Library

```python
import numpy as np
import pandas as pd
```

### 2. Load Dataset

```python
ulasan = [
    "aplikasi sangat bagus dan cepat",
    "layanan jelek sekali, sangat kecewa",
    "lumayan oke lah untuk harga segini",
    "sering error dan force close terus",
    "biasa saja tidak ada yang spesial",
]
df = pd.DataFrame({"teks": ulasan})
df.head()
```

### 3. Fungsi Programmatic

```python
# Aturan 1: Cari kata positif (1)
def lf_positif(teks):
    return 1 if any(w in teks for w in ["bagus", "cepat", "oke"]) else -1
```

```python
# Aturan 2: Cari kata negatif (0)
def lf_negatif(teks):
    return 0 if any(w in teks for w in ["jelek", "kecewa", "error"]) else -1
```

```python
# Gabungkan aturan
def auto_label(teks):
    if lf_positif(teks) == 1:
        return 1
    if lf_negatif(teks) == 0:
        return 0
    return -1  # Abstain / ragu-ragu (-1)
```

### 4. Tampilkan Data

```python
df["weak_label"] = df["teks"].apply(auto_label)
df[["teks", "weak_label"]]
```
