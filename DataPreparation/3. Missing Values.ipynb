### 1. Load Library

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer, SimpleImputer
```

### 2. Load Dataset

```python
df = pd.DataFrame(
    {
        "usia": [25, 30, np.nan, np.nan, 22],
        "gaji": [5000000, np.nan, 7000000, 12000000, 4500000],
        "kota": ["Jakarta", "Bandung", np.nan, "Jakarta", "Surabaya"],
    }
)
df.head()
```

### 3.1. Check Missing Value

```python
df.isna().sum()
```

### 3.2 Hapus Missing Value Value

```python
df_drop = df.dropna()
df_drop
```

### 3.3. Median untuk numerik

```python
imputer_num = SimpleImputer(strategy="median")
df["usia"] = imputer_num.fit_transform(df[["usia"]])
df
```

### 3.4. Imputation Prediktif

```python
knn_imp = KNNImputer(n_neighbors=2)
df_knn = df
df_knn[["usia", "gaji"]] = knn_imp.fit_transform(df_knn[["usia", "gaji"]])
df_knn
```
