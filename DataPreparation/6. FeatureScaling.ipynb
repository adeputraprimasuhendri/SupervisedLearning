### 1. Load Library

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

### 2. Load Dataset

```python
data = {
    "usia": [20, 25, 30, 45, 50],  # Rentang puluhan
    "gaji": [
        3_000_000,
        5_000_000,
        8_000_000,
        15_000_000,
        25_000_000,
    ],  # Rentang jutaan
}
df = pd.DataFrame(data)
df
```

### 3.1. Min-Max Normalization (Skala [0, 1])

```python
minmax_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(df),
    columns=["usia_minmax", "gaji_minmax"],
)
df_minmax = pd.concat([df, df_minmax], axis=1)
df_minmax
```

### 3.2. Standardization / Z-Score (Mean = 0, Std = 1)

```python
std_scaler = StandardScaler()
df_std = pd.DataFrame(
    std_scaler.fit_transform(df),
    columns=["usia_zscore", "gaji_zscore"],
)
df_std = pd.concat([df, df_std], axis=1)
df_std
```
